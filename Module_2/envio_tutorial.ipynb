### What is blockchain indexing

This is the process of extracting, organizing and storing blockchain  data in a structured way so it can be queried quickly

Doc: https://docs.envio.dev/

example: https://docs.envio.dev/blog/how-to-build-visualizers-and-dashboards-on-monad-using-envio

```python
# graphql endpoint: http://localhost:8080/v1/graphql
import requests

url = "http://localhost:8080/v1/graphql"

# query to fetch transfers
query = """
query MyQuery {
  NonfungiblePositionManager_IncreaseLiquidity(limit: 1900) {
    db_write_timestamp
    id
    tokenId
    liquidity
    amount0
    amount1
    
  }
}
"""

# If authentication is needed, add it to headers (Envio often works without auth for dev endpoints)
headers = {
    "Content-Type": "application/json"
}

# Send request
response = requests.post(url, json={"query": query}, headers=headers)

# Handle response
if response.status_code == 200:
    data = response.json()
    print(data)  # full JSON
else:
    print(f"Query failed with status {response.status_code}")
    print(response.text)
```

```python
# length of the response data
print(f"Number of records returned: {len(data.get('data', {}).get('NonfungiblePositionManager_IncreaseLiquidity', []))}")  # Adjust based on your query structure   
```
