#### Scrap Data from Coingecko using BeautifulSoup

```python
from bs4 import BeautifulSoup
from selenium import webdriver
import requests
import time

# Get the web page
driver = webdriver.Chrome()
driver.get("https://www.coingecko.com/en")
time.sleep(5)  # Wait for the page to load completely
page = requests.get("https://www.coingecko.com/en")
```

### Reading the page

```python
soup = BeautifulSoup(page.content, "html.parser")
print(soup.prettify()[:200])  # Print the formatted HTML content

driver.quit()  # Close the browser after scraping
```

#### Scrap CoinMarketCap

```python
url = "https://coinmarketcap.com/currencies/bitcoin/historical-data/?start=20200101&end=20231231"

# Send HTTP GET request to the URL
response = requests.get(url)
soup = BeautifulSoup(response.content, "html.parser")

# Find the table containing historical data
# The class name is now "h7vnx2-2 czTsgW cmc-table" without the extra space
table = soup.find("table", {"class": "h7vnx2-2 czTsgW cmc-table"})

# Extract the top 10 rows of the table
if table:
    rows = table.find_all("tr")[1:11]  # Using find_all is more robust
    
    for row in rows:
        # Extract the columns in each row
        name = row.find("p", {"class": "sc-4984d93-0"}).text
        price = row.find("td", {"class": "sc-16z8icm-0"}).text
        market_cap = row.find_all("td", {"class": "sc-7bc61c98-0"})[0].text
        print(f"Coin: {name}, Price: {price}, Market Cap: {market_cap}")
else:
    print("Error: Table not found. The class name may have changed.")
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

def scrape_dexscreener(url):
    """
    Scrapes the main price from a Dexscreener page using Selenium.
    """
    
    # Configure Chrome to run in headless mode (no browser window)
    options = webdriver.ChromeOptions()
    options.add_argument('--headless')
    options.add_argument('--disable-gpu')
##@#@WW###
# 
# 
# 
# 
#  ff        aEDFVBV
        
        
        
        
        
        dcddd'--no-sandbox')

    # Initialize the Chrome WebDriver
    driver = webdriver.Chrome(options=options)
    
    try:
        print(f"Navigating to {url}...")
        driver.get(url)
        
        # Use WebDriverWait to wait until the price element is visible
        # The class name is an example and may need to be updated.
        price_selector = "span.chakra-heading" 
        
        wait = WebDriverWait(driver, 10)
        price_element = wait.until(
            EC.presence_of_element_located((By.CSS_SELECTOR, price_selector))
        )
        
        # Extract and print the price
        price = price_element.text
        print(f"✅ Successfully scraped price: {price}")
        
    except Exception as e:
        print(f"❌ An error occurred: {e}")
        
    finally:
        # Always close the browser
        print("Closing the browser.")
        driver.quit()

# Example URL for a specific trading pair (e.g., a popular one)
# You need to replace this with the specific URL you want to scrape.
dexscreener_url = "https://dexscreener.com/base/0x0d507b51b392b51664d4b17f8a7e58b855581165"
scrape_dexscreener(dexscreener_url)
```
